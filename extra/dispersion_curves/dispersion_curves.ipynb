# Dispersion curves of CEN2, CEN4 and UBS

Theoretical phase speed, group speed and damping (space discretisation only).

```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(1e-3, np.pi, 400)
S2 = np.sin(theta)
S4 = 4/3*np.sin(theta) - 1/6*np.sin(2*theta)
D  = (1 - np.cos(theta))**2/3   # UBS damping
```

```python
fig, axs = plt.subplots(1, 3, figsize=(14, 4))
axs[0].plot(theta, S2/theta, label='CEN2')
axs[0].plot(theta, S4/theta, label='CEN4 = UBS')
axs[0].set_title('$c_p/c_0$')
axs[1].plot(theta, np.cos(theta), label='CEN2')
axs[1].plot(theta, 4/3*np.cos(theta) - 1/3*np.cos(2*theta), label='CEN4 = UBS')
axs[1].set_title('$c_g/c_0$')
axs[2].plot(theta, -D, label='UBS')
axs[2].set_title('$\\omega_i \\Delta x / c_0$')
for ax in axs:
    ax.axhline(0, color='k', lw=.5); ax.set_xlabel(r'$\theta = k\Delta x$'); ax.legend(); ax.grid(alpha=.3)
plt.tight_layout(); plt.show()
```

```python
# Time stepping correction (RK3), Courant number C
C = 0.09
y = C*S4
cp_rk3 = np.arctan2(y - y**3/6, 1 - y**2/2)/(C*theta)
print('max |cp_rk3 - cp_space| =', np.abs(cp_rk3 - S4/theta).max())
```
